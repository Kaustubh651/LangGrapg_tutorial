```python
import os
from langgraph.graph import StateGraph, START,END
from openai import OpenAI
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
# creating the state

class LLMState(TypedDict):
    question:str
    answer:str
```

```python
def llm_qa(state: LLMState)->LLMState:

    ## extract the question from the state
    question = state['question']

    ## form a prompt
    prompt = f"Answer the following question: {question}"

    ## ask that question to the LLM and get the answer
    client = OpenAI(
        base_url="https://router.huggingface.co/v1",
        api_key=os.environ["HF_TOKEN"],
    )
    completion = client.chat.completions.create(
            model="Qwen/Qwen3-8B",
            messages=[
                {
                    "role": "user",
                    "content": prompt
                },
            ],
    )

    ## store the answer back in the state and return it
    state["answer"] = completion.choices[0].message.content
    
    return state
```

```python
# creating the graph

graph = StateGraph(LLMState)

# add node to your graph
graph.add_node("llm_qa",llm_qa)


# add edges to your graph
graph.add_edge(START, "llm_qa")
graph.add_edge("llm_qa",END)


# execute the graph
workflow= graph.compile()
```

```python
## execute the graph with a question
initial_state = {'question': 'What is alien life?'}


final_state = workflow.invoke(initial_state)

print(final_state['answer'])
```
